##Задание 12. Шифр Цезаря

```python
#Реализуйте шифрование текста сдвигом на k позиций:
#работайте только с латинскими буквами (a–z, A–Z), остальные символы не меняйте;
#для сдвига используйте ord()/chr() и арифметику по модулю 26;
#напишите две переменные-функции (пока без def): зашифрованный и расшифрованный текст.
#Проверьте на короткой строке "Hello, World!" с k = 3.

s = "Hello, World!"
k = 3

encrypted = ""

for char in s:
    if char.isalpha():
        if char.isupper():
            encrypted += chr((ord(char) - ord('A') + k) % 26 + ord('A'))
        else:
            encrypted += chr((ord(char) - ord('a') + k) % 26 + ord ('a'))
    else:
        encrypted += char

print("Зашифрованный:", encrypted)

decrypted = ""

for char in encrypted:
    if char.isalpha():
        if char.isupper():
            decrypted += chr((ord(char) - ord('A') - k) % 26 + ord('A'))
        else:
            decrypted += chr((ord(char) - ord('a') - k) % 26 + ord('a'))

    else:
        decrypted += char

print("Расшифрованный:", decrypted)




```

```python
#Задание 16. Топ-3 частых слова без словаря
#Через str.count() и цикл for по уникальным словам:
#получите список уникальных слов (можно через set(text.split()) — тоже аванс);
#для каждого посчитайте частоту;
#найдите 3 самых частых слова, выводя их с частотой.
#Это неоптимально по скорости, но показывает, как обходиться без dict.

text = input("Введите текст: ")

words = text.lower().split()
unique_words = set(words)

result = []

for word in unique_words:
    count = text.lower().count(word)
    result.append([word, count])

result.sort(key=lambda x: x[1], reverse=True)

print("Топ-3 самых частых слова:")

for item in result[:3]:
    print(item[0], "-", item[1])
```

```python
#Реализуйте простейший Pig Latin:
#если слово начинается с согласной — перенесите первую букву в конец и
#добавьте "ay": "python" → "ythonpay";
#если с гласной — добавьте "way" в конец: "apple" → "appleway";
#обработайте первые 20 слов из текста и выведите результат.
text = input("Введите текст: ")

words = text.split()

vowels = "aeiou"

for word in words[:20]:
    if word[0].lower() in vowels:
        result = word + "way"
    else:
        result = word[1:] + word[0] + "ay"

    print(result)
```

##Задание 19. Оберните всё в функции

```python
#Программирование на языке Python, Киреев В.С., к.т.н., доцент
#Перепишите логику в функции с аннотациями и docstring: def download_text(url: str, path: str) -> None: ... def read_text(path: str) -> str: ... def clean_words(text: str) -> list[str]: ... def find_capitalized(text: str) -> list[str]: ... def is_palindrome(text: str) -> bool: ... def caesar(text: str, shift: int) -> str: ... def char_stats(text: str) -> None: ... def pig_latin(word: str) -> str: ... def parse_url(url: str) -> tuple[str, str, str, str]: ... def main() -> None: ...
#Соберите всё в main() и вызовите через if __name__ == "__main__":.

import urllib.request


def download_text(url: str, path: str) -> None:
    """Скачивает текст по URL и сохраняет его в файл."""
    urllib.request.urlretrieve(url, path)


def read_text(path: str) -> str:
    """Читает текст из файла и возвращает его содержимое."""
    with open(path, "r", encoding="utf-8") as file:
        return file.read()


def clean_words(text: str) -> list[str]:
    """Возвращает список слов, приведённых к нижнему регистру."""
    words = text.split()
    result = []

    for word in words:
        word = word.strip(".,!?;:\"()[]")
        if word:
            result.append(word.lower())

    return result


def find_capitalized(text: str) -> list[str]:
    """Находит слова, начинающиеся с заглавной буквы."""
    words = text.split()
    result = []

    for word in words:
        if word and word[0].isupper():
            result.append(word)

    return result


def is_palindrome(text: str) -> bool:
    """Проверяет, является ли текст палиндромом."""
    text = text.lower()
    return text == text[::-1]


def caesar(text: str, shift: int) -> str:
    """Шифрует текст шифром Цезаря."""
    result = ""

    for char in text:
        if char.isalpha():
            start = ord("A") if char.isupper() else ord("a")
            result += chr((ord(char) - start + shift) % 26 + start)
        else:
            result += char

    return result


def char_stats(text: str) -> None:
    """Выводит количество букв, цифр, пробелов и других символов."""
    letters = 0
    digits = 0
    spaces = 0
    other = 0

    for char in text:
        if char.isalpha():
            letters += 1
        elif char.isdigit():
            digits += 1
        elif char.isspace():
            spaces += 1
        else:
            other += 1

    print("Буквы:", letters)
    print("Цифры:", digits)
    print("Пробелы:", spaces)
    print("Другие символы:", other)


def pig_latin(word: str) -> str:
    """Переводит одно английское слово в простой Pig Latin."""
    vowels = "aeiou"

    if not word:
        return word

    if word[0].lower() in vowels:
        return word + "way"

    return word[1:] + word[0] + "ay"


def parse_url(url: str) -> tuple[str, str, str, str]:
    """Извлекает из URL протокол, домен, имя файла и расширение."""
    parts = url.split("://")

    protocol = parts[0]
    rest = parts[1]

    parts = rest.split("/")

    domain = parts[0]
    filename = parts[-1]

    dot = filename.rfind(".")
    extension = filename[dot:]

    return protocol, domain, filename, extension


def main() -> None:
    """Запускает основные примеры работы программы."""

    text = "Hello world! Python is a great language."

    print("Исходный текст:")
    print(text)

    print("\nОчищенные слова:")
    print(clean_words(text))

    print("\nСлова с заглавной буквы:")
    print(find_capitalized(text))

    print("\nПалиндром:")
    print(is_palindrome("А роза упала на лапу Азора"))

    print("\nШифр Цезаря:")
    print(caesar("Hello World", 3))

    print("\nСтатистика символов:")
    char_stats(text)

    print("\nPig Latin:")
    print(pig_latin("python"))
    print(pig_latin("apple"))

    print("\nРазбор URL:")
    url = "https://www.gutenberg.org/files/74/74-0.txt"
    protocol, domain, filename, extension = parse_url(url)

    print("Протокол:", protocol)
    print("Домен:", domain)
    print("Файл:", filename)
    print("Расширение:", extension)


if __name__ == "__main__":
    main()
```
